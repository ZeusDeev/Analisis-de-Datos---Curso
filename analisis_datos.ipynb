### Series y DataFrame con pandas.

Una serie es una estructura de datos unidimencional que permite almacenar cualquier tipo de datos.

Por otro lado los dataframe son estructura de datos bidimencionales que pueden contener multiples columnas y filas de datos y ambos pueden ser creados apartir de listas,diccionarios o arreglos de numpy.

```python
import pandas as pd
```

```python
#Series
numeros = [1,2,3,4]
series = pd.Series(numeros)
print(series)
```

 ## DataFrame con Pandas:


 En un **dataframe** podemos crear desde un diccionario, colocando primeramente el indice de la tabla, es decir el encabezado de nuestra columna, seguido de una lista para colocar nuestros valores.

```python
#DataFrame

datos = {
    'Nombre':['Juan','Maria','Pedro','Abraham'],
    'Edad':[15,20,12,4],
    'Ciudad':['Lima','Arequipa','Lima','Carrillo Puerto']
}


#Genera un DataFrame apartir de un diccionario.

df = pd.DataFrame(datos)
print(df)
```

###Exporta Dataframe como archivo csv.

```python
df.to_csv('datos.csv')
```

### Importa un DataFrame

```python
df = pd.read_csv('datos.csv', index_col = 0)
# df = df.drop(df.columns[0], axis=1) #con df.drop, eliminamos la columna 0 que salia por defecto, aplicando en todos los campos por medio de axis=1;
df
```

## **Manipulación de datos con pandas**

```python
#Seleccionar una columna

nombres = df['Nombre']
nombres
```

```python
#Seleccionar una o más columna

fisrt_column = df[['Nombre','Ciudad']]
fisrt_column
```

```python
#Filtrar por indice
df.loc[3]
```

```python
#Filtrar por condición
df[df['Edad'] < 5]
```

```python
#Filtrar con Query

# df.query('Ciudad == "Lima"')
df.query('Edad > 10')
```

```python
# Filtra por nombre especificos usando isin

df[df['Nombre'].isin(['Juan','Abraham','Maria'])]
```

```python
# Filtrar por ciudad usando isin

df[df['Ciudad'].isin(['Lima'])]
```

```python
# Filtrado por funcion considerando solo 5 caracteres de nombre

def filtrado_caracter(nombre):
  return len(nombre) == 5

df[df['Nombre'].apply(filtrado_caracter)]
```

```python
# Crea una funcion que me permita encontrar las personas con edad menor a 15,
# ademas de considerar que los caracteres en nombres sean menor o igual a 5.

def encontrar_persona(edad, nombre):
  return (edad <= 15) & (nombre.str.len() <= 5)



filter = df[encontrar_persona(df['Edad'], df['Nombre'])]
filter
```

```python
# Filtrar por edades etre 25 y 35 años

df[df['Edad'].between(12,20)]
```
