# **Pandas - Tratamiento de Datos**

```python
import pandas as pd
import numpy as np

# Creación del diccionario
data = {
    'Nombre': ['Mateo', 'Esteban', 'Leonardo', 'Wily', 'Carlos','Mateo','Julio','Abraham','Ramiro','Casimiro','Manuel','Jose','Marcos','Francisco','Alex'],
    'Apellido': ['Perez', None, 'Dicaprio', 'Wonka', 'Pereira',None,'Mato','Cach','Alfalfa','Socorro','Maquiavelo',None,'Estrada', 'Luna', None],
    'Edad': [20, 32, np.nan, 25, np.nan, 23, 60, 34, 20, 29,50, np.nan, 18, 25, 42],
    'Ciudad': ['CDMX', None, None, 'CDMX', 'Delicias', 'Guadalajara', None, 'Matamoros', 'Los Cabos', 'Tenochtitlan', 'Jalisco', None, 'Guanajuato', 'Tuxtla', 'Puebla']
}

# Convertimos nuestro diccionario a un dataframe
personal_data = pd.DataFrame(data)
```

```python
# Exportación del dataframe
personal_data.to_csv('personal_data.csv')
```

```python
# Importacion del dataframe
personal_data = pd.read_csv('personal_data.csv')
```

```python
# Eliminamos columna extra que sale unnamed
personal_data.drop(personal_data.columns[0], axis=1)
```

```python
# Rellenar valores Faltantes con fillna
personal_data_fillna = personal_data.fillna({
    'Edad': personal_data['Edad'].mean(), # si edad esta vacio, accedemos a nuestro df en la columna edad para cambiarlo por la media
    'Ciudad': 'Desconocida' # si ciudad esta vacia, se cambia por desconocida
})

personal_data_fillna #imprime
```

```python
# Si notamos que apesar de rellenar nuestros valores faltantes o sabemos que puede
# afectar a nuestra muestra, podemos aplicar los siguiente, usando dropna:

persona_data_dropna = personal_data.dropna() #elimina todos los datos con valores nulos, esto aplica para toda la fila sin importar solo sea un valor faltante
persona_data_dropna #imprime

```

```python
# Remplazar valores especificos de alguna columna usando replace

# Igual, otra manera disponible para rellenar valores nulos, puede ser con
# un diccionario, como con fillna pero en este caso seleccionando un tipo especifico
# cono np.nan, None, etc; Seguido de la palabra que deseamos remplazar

personal_data_replace = personal_data.replace({
    'Ciudad': {None: 'Desconocida'}, #si ciudad contien el valor None, cambia a 'Desconocida'
    'Edad': {np.nan: personal_data['Edad'].mean()} #si edad tiene valores NaN, accedemos a nuestro df en la columna edad para cambiarlo por la media
})

personal_data_replace #imprime
```

```python
# Rellenado de datos faltantes por medio de la interpolacion o bien usando interpolate

personal_data_interpolate = personal_data.copy() #Hacemos una copia de nuestra anterior data, para rellenar valores nulos, guiandose de lo anterior
personal_data_interpolate['Edad'] = personal_data['Edad'].interpolate() # seleccionamos la columna con valores nulos y despues seleccionamos nuestro data origial, y aplicamos interpolate

personal_data_interpolate #imprime
```

# **Otra maneras de resolver datos duplicados**

```python
# Generarmos un df con datos duplicados


df_duplicado = {
    'Nombre': ['Mateo', 'Juan', 'Mateo', 'Zeus', 'Zeus'],
    'Edad' : [22, 24, 22, 20, 20],
    'Ciudad' : ['CDMX', 'Cancun', 'CDMX', 'Monterrey', 'Monterrey']
}

#pasamos de diccionario a dataframe
df_duplicado = pd.DataFrame(df_duplicado)

#imprimimos
df_duplicado
```

```python
# Elimanos datos duplicados con drop_duplicates()

# cramos un df nuevo no duplicado, pasamos nuestro df duplicado para asignarle drop_duplicates() para cometer la solucion de duplicacion
df_no_duplicado = df_duplicado.drop_duplicates()
df_no_duplicado #imprimimos
```

## **Un tip es que igual si deseamos podemos cambiar el titulo de nuestras columnas**

```python
# Cambiamos el titulo de nuestras columnas de español a ingles usando rename.

df_renombrado =  df_no_duplicado.rename(columns={  #creamos un df nuevo, pasando nuestro df anterior o actual, para usar lrename seguido de columns y renombrar por medio de diccionario
    'Nombre':'Name',
    'Edad':'Age',
    'Ciudad':'City'
})

df_renombrado #imprime
```

## **De igual manera, podemos ordenar nuestras columnas**

```python
# Ordenamiento de columnas

columnas_ordenadas = ['City', 'Age', 'Name'] #ordenamos como queremos las columnas
df_ordenado = df_renombrado[columnas_ordenadas] #Despues creamos un df ordenado por medio de nuestro df actual asigandole las columans ordenadas

df_ordenado #imprimimos
```
