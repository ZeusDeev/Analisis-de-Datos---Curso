# **Agrupacion de datos**

```python
import pandas as pd
import numpy as np


# Creación del diccionario
data = {
    'Nombre': ['Mateo', 'Esteban', 'Leonardo', 'Wily', 'Carlos','Mateo','Julio','Abraham','Ramiro','Casimiro','Manuel','Jose','Marcos','Francisco','Alex'],
    'Apellido': ['Perez', None, 'Dicaprio', 'Wonka', 'Pereira',None,'Mato','Cach','Alfalfa','Socorro','Maquiavelo',None,'Estrada', 'Luna', None],
    'Edad': [20, 32, np.nan, 25, np.nan, 23, 60, 34, 20, 29,50, np.nan, 18, 25, 42],
    'Ciudad': ['CDMX', None, 'Guadalaja', 'CDMX', 'Delicias', 'Guadalajara', None, 'Matamoros', 'Los Cabos', 'Tenochtitlan', 'Jalisco', 'Los Cabos', 'Guanajuato', 'Tuxtla', 'Puebla'],
    'Puntuacion': [80, 90, 85, 78, 50, 60, 45, 88, 95, 74, 67, 85, 34, 56, 99]
    }

data = pd.DataFrame(data)
data
```

# **Limpieza de datos ( conceptos aplicados anteriormente)**

Antes de iniciar con el agrupamiento limpiaremos los datos, datos que son compuestos como nulos, usa la opcion que mas te parezca viable

```python
data_clean = data.replace({
    'Edad': {np.nan : data['Edad'].mean()},
    'Ciudad': {None : data['Ciudad'].mode()},
    'Apellido' : {None : 'Desconocido'}
})

data_clean
```

## Este ejercicio de agrupación nos regresa un diccionario según la ciudad donde viven las personas, por ejemplo, el índice 0 y 3 viven en CDMX

```python
# Agrupa los datos por la columna 'Ciudad'
grouped = data.groupby('Ciudad')

# Imprime los grupos resultantes por 'Ciudad'
print(grouped.groups)

```

## **Otra manera puede ser con el agg:**

Con el groupby agruparas los registros por ciudades , con la funcion agg y la opcion "count" contaras cada registro que exista por ciudad

```python
data.groupby("Ciudad")["Ciudad"].agg(["count"])
```
